```python
import requests #for get request
import pandas as pd #pandas
import numpy as np #module
import re #regex
from datetime import datetime #datetime objects
from bs4 import BeautifulSoup #prettify's our content
import json #needed for google API
import os #needed for file reading
import gmaps #for our heatmaps
import gmaps.datasets
import matplotlib.pyplot as plt #for plotting
from sklearn import linear_model #for linear regression
from sklearn.preprocessing import PolynomialFeatures #polynomial regression
```

```python
year = 2025
months = ['01','02','03','04','05','06','07','08','09','10','11','12']
# SCRAPE EVERY MONTHS DATA
df_lst = [] # LST OF EACH MONTHS DATA
for month in months:
    req = requests.get(f"https://umpd.umd.edu/statistics-reports/daily-crime-and-incident-logs/{year}/{month}") #GET REQUEST
    print(f"Month {month} - Status Code: {req.status_code}")
    
    if req.status_code != 200:
        print(f"Failed to retrieve data for month {month}")
        continue
    
    chicken_soup = BeautifulSoup(req.content, "html.parser")
    table_list = str(chicken_soup.find("table")) #FIND THE TABLE
    
    if table_list == 'None':
        print(f"No table found for month {month}")
        continue
    
    try:
        df_lst.append(pd.read_html(table_list, flavor='html5lib',header=0)[0]) #CREATE AND APPEND THE DF FOR THE CURR MONTH TO THE DF_LIST
    except ValueError as e:
        print(f"Error parsing table for month {month}: {e}")
        continue

if df_lst:
    df = pd.concat(df_lst) #CONCATENATE ALL THE DFs IN THE LIST INTO ONE LARGE DF
    df.head()
else:
    print("No data was successfully scraped")
```

```python
# after concatenating all months
ADDRESS_COL = "Unnamed: 5"

df = df.reset_index(drop=True)
df["LOCATION"] = ""

for idx in range(1, len(df), 2):  # the rows that hold only the address
    df.at[idx - 1, "LOCATION"] = df.at[idx, ADDRESS_COL]

df = df.iloc[::2].reset_index(drop=True)
df = df.drop(columns=[ADDRESS_COL])
```

```python
import os
import json
import time
import pandas as pd
import requests
from dotenv import load_dotenv

# ---------- LOAD .env ----------
load_dotenv()

# Try either GOOGLE_MAPS_API_KEY or GOOGLE_API_KEY from your .env
GEOCODING_API_KEY = os.getenv("GOOGLE_MAPS_API_KEY") 

if GEOCODING_API_KEY is None:
    raise ValueError(
        "No Geocoding API key found. Set GOOGLE_MAPS_API_KEY or GOOGLE_API_KEY in your .env file."
    )

# ---------- GOOGLE GEOCODING HELPER ----------
def geocode_with_google(location: str):
    """
    Use Google Geocoding API to get (lat, lng) for a location string.
    Returns (lat, lng) or (None, None) if it fails.
    """
    if not isinstance(location, str) or not location.strip():
        return None, None

    # Optional: bias the search around College Park, MD
    # so vague strings like "Stamp Student Union" still resolve nicely
    address = f"{location}, College Park, MD"

    url = "https://maps.googleapis.com/maps/api/geocode/json"
    params = {
        "address": address,
        "key": GEOCODING_API_KEY,
    }

    try:
        resp = requests.get(url, params=params)
        resp.raise_for_status()
        data = resp.json()

        if data.get("status") != "OK" or not data.get("results"):
            print(f"Geocoding failed for '{location}': {data.get('status')}")
            return None, None

        loc_data = data["results"][0]["geometry"]["location"]
        lat = loc_data.get("lat")
        lng = loc_data.get("lng")
        return lat, lng

    except Exception as e:
        print(f"Geocoding failed for '{location}': {e}")
        return None, None


# ---------- APPLY TO YOUR DATAFRAME ----------
# Assuming your DataFrame is named df and has a 'LOCATION' column

# 1. Get unique locations
unique_locations = (
    df["LOCATION"]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

location_to_coords = {}

# 2. Call Geocoding API once per unique location
for loc in unique_locations:
    lat, lng = geocode_with_google(loc)
    location_to_coords[loc] = (lat, lng)
    time.sleep(0.1)  # be a bit gentle with the API; adjust as needed

# 3. Map back to the dataframe as new columns
def lookup_lat(loc):
    loc = str(loc).strip()
    return location_to_coords.get(loc, (None, None))[0]

def lookup_lng(loc):
    loc = str(loc).strip()
    return location_to_coords.get(loc, (None, None))[1]

df["Latitude"] = df["LOCATION"].apply(lookup_lat)
df["Longitude"] = df["LOCATION"].apply(lookup_lng)

# 4. Quick check
df[["LOCATION", "Latitude", "Longitude"]].head()
```
